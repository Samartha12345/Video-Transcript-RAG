# YouTube transcript question answering

Use the kernel from the project environment. The sample uses a local transcript. The first model run requires internet access to download the weights.

```python
from pathlib import Path
from youtube_rag.transcripts import load_transcript
from youtube_rag.retrieval import index_transcript
from youtube_rag.generation import answer_question

root = next(p for p in (Path.cwd(), *Path.cwd().parents) if (p / "examples/sample_transcript.txt").exists())
text, source = load_transcript(None, root / "examples/sample_transcript.txt")
# For a live video, replace the preceding line with:
# text, source = load_transcript("Gfr50f6ZBvo", None)
print(source, len(text), "characters")
print(text[:400])
```

## Build the index

MiniLM encodes overlapping passages. Chroma stores the vectors for similarity search. Run this cell once per transcript.

```python
retriever = index_transcript(text, source)
```

## Retrieve evidence

Try another question and inspect the passage text before generating an answer.

```python
question = "How often does the vibration sensor collect readings?"
passages = retriever.invoke(question)
for doc in passages:
    print(f"Passage {doc.metadata['passage']} ({doc.metadata['source']})")
    print(doc.page_content, end="\n\n")
```

## Generate an answer

Qwen receives the question and the passages on CPU. Compare its answer with the passages above. Try an unanswered question such as “What was the motor serial number?” to explore the limits of the prompt.

```python
print(answer_question(question, passages))
```
